# YOLO Model Training — Infrastructure Defect Detection

Fine-tune YOLOv8 on a custom infrastructure defect dataset.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('.').resolve().parent))

from src.detection.yolo_detector import YOLODetector

DATASET_YAML = '../models/dataset.yaml'
detector = YOLODetector()
```

```python
# Train YOLO on custom dataset
# Ensure models/dataset.yaml points to your labeled data

results = detector.train(
    data_yaml=DATASET_YAML,
    epochs=100,
    imgsz=640,
    batch=16,
)
print(results)
```

```python
# Evaluate on a test image
test_image = '../data/raw/sample.jpg'  # Replace with your test image

if Path(test_image).exists():
    detections = detector.detect(test_image)
    print(f'Detections: {len(detections)}')
    for d in detections:
        print(f"  {d['class']}: {d['confidence']:.2%} ({d['severity']})")
else:
    print('Add a test image to data/raw/sample.jpg')
```

```python
# Train predictive maintenance model
from src.models.predictive_maintenance import PredictiveMaintenanceEngine

sample_records = [
    {'asset_id': 1, 'class': 'crack', 'confidence': 0.9, 'severity': 'critical', 'detected_at': '2026-01-15'},
    {'asset_id': 1, 'class': 'rust', 'confidence': 0.7, 'severity': 'high', 'detected_at': '2026-02-10'},
    {'asset_id': 2, 'class': 'corrosion', 'confidence': 0.5, 'severity': 'medium', 'detected_at': '2026-03-01'},
]
labels = ['critical', 'high']

engine = PredictiveMaintenanceEngine()
train_result = engine.train(sample_records, labels)
print(train_result)

predictions = engine.predict_risk(sample_records)
for p in predictions:
    print(f"Asset #{p['asset_id']}: {p['risk_level']} — {p['recommended_action']}")
```
